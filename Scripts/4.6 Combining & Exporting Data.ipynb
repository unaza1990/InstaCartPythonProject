# Import Libraries

# Import Dataframes

```python
# Import dataset orders_products_combined.pkl
df_ords_prods_combined = pd.read_pickle(os.path.join(path, 'Data', 'Prepared Data', 'orders_products_combined.pkl'))
```

```python
# Import the cleaned products data
df_products = pd.read_csv(os.path.join(path, 'Data', 'Prepared Data', 'products_cleaned.csv'))
```

# Combine the dataframe

```python
df_ords_prods_merge = df_ords_prods_combined.merge(df_products, on='product_id', how='inner')
```
